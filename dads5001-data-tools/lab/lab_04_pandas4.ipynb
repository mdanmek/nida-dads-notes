#### Left join (left outer join):

The new merged DataFrame <u>includes all rows from the left DataFrame</u>, while discarding rows from the right DataFrame that don’t have a match in the key column of the left DataFrame.

```python
# Alternative 1
df_left = pd.merge( df1_pokemon, df2_move, left_on='type2', right_on='Type', how='left' )

# Alternative 2
#df_left = df1_pokemon.merge( df2_move, left_on='type2', right_on='Type', how='left' )

# Display
df_left
```

#### Right join (right outer join):

<u>All rows from the right DataFrame will be retained</u>, while rows in the left DataFrame without a match in the key column of the right DataFrame will be discarded.

#### Outer join (full outer join):

<u>All rows from both DataFrames will be present in the new DataFrame</u> as this outer join returns all rows which either have a match in the left or right DataFrames.

```python
# Alternative 1
df_outer = pd.merge( df1_pokemon, df2_move, left_on='type2', right_on='Type', how='outer' )

# Alternative 2
#df_outer = df1_pokemon.merge( df2_move, left_on='type2', right_on='Type', how='outer' )

# Display
df_outer
```

#### Cross join (cartesian product):

Create the cartesian product from both DataFrames while preserving the order of the left keys

```python
# Alternative 1
df_cross = pd.merge( df1_pokemon, df2_move, how='cross' )

# Alternative 2
#df_cross = df1_pokemon.merge( df2_move, how='cross' )

# Display
df_cross
```

# 2. Pandas: Combine dataframes by stacking them along an axis

- With `pd.merge()` (and `pd.join()`), we can expect the resulting dataset to have rows from the parent datasets mixed in together, <u>often based on some commonality</u>.

- With `pd.concat()`, <u>our datasets are just stacked together along an axis</u> — either the row axis or column axis. It is a simpler way to combine datasets and is often used to form a single, larger set to do additional operations on.
  - In `np.concatenate()`, dimensions must match along the non-concatenated axis to avoid errors.
  - In `pd.concat(...,join)`, <u>columns are automatically aligned based on their labels</u>.
    - `join='outer'` (default): A union of the labels on the non-concatenation axis (= include all labels from both DataFrames). Fill the missing row or column with NaN.
    - `join='inner'`: An intersection of the labels on the non-concatenation axis (= only include labels that are present in all of the DataFrames). Any rows or columns with a label that is not shared are dropped.

- `pd.merge(...,how)` vs. `pd.concat(...,join)`:
  - There commands are generally not interchangeable, although there are specific edge cases where they might produce a similar final result.
  - While both align data, `pd.concat()` only aligns on labels, whereas `pd.merge()` primarily aligns on values but can also be explicitly told to align on the label.
  - When two tables don't share the same label but have a common column, `pd.merge()` is the only correct tool.

```python
# Choose only water pokemons
df_water = pokemon_df.loc[ pokemon_df['type1']=='water', ['pokedex_number','name','type1','type2','abilities'] ]

# shape=(114,5)
df_water = df_water.reset_index(drop=True)
df_water
```

```python
# Choose only fire pokemons
df_fire = pokemon_df.loc[ pokemon_df['type1']=='fire', ['type1','type2','pokedex_number','name'] ]

# shape=(52,4)
df_fire = df_fire.reset_index(drop=True)
df_fire
```

## 2.1 Concat along axis=0 (row)

Stack DataFrames vertically
- `join='outer'` (default): Include all columns from all dataframes. For any row with missing values, fill them with NaN.
- `join='inner'`: Include only columns that appear in all dataframes, drop the other columns.

### `join='outer'`

```python
# join='outer': Include all columns from all dataframes
df = pd.concat( [df_water, df_fire] )    # default:axis=0, join='outer'
display(df.info())
display(df)
```

```python
# join='outer': Include all columns from all dataframes
# Same results as the above cell but with the different order of columns
df = pd.concat( [df_fire, df_water] )    # default:axis=0, join='outer'
display(df.info())
display(df)
```

```python
# join='outer': Include all columns from all dataframes
# Set ignore_index=True to reset the existing row labels
df = pd.concat( [df_fire, df_water], ignore_index=True  )    # default:axis=0, join='outer', ignore_index=False
display(df.info())
display(df)
```

### `join='inner'`

```python
# join='inner': Include only columns that appear in all dataframes
df = pd.concat( [df_water, df_fire], join='inner' )    # default:axis=0
display(df.info())
display(df)
```

## 2.2 Concat along axis=1 (column)

Stack DataFrames horizontally
- `join='outer'` (default): Include all rows from all dataframes. For any column with missing values, fill them with NaN.
- `join='inner'`: Include only rows that appear in all dataframes, drop the other rows.

```python
# df_water.shape = (114,5)
# df_fire.shape = (52,4)
df = pd.concat( [df_water, df_fire], axis=1 )    # default:axis=0, join='outer'
df
```

```python
# df_water.shape = (114,5)
# df_fire.shape = (52,4)
df = pd.concat( [df_fire, df_water], axis=1 )    # default:axis=0, join='outer'
df
```

## 2.3 Get rid of the automatic column alignment

Make `pd.concat()` behave like `np.concatenate()` --- stacking the data without considering column names

```python
# Manually reset column labels to numerical indices
# set_axis() renames columns to [0, 1, 2, ...], removing the original labels.
df_water_reset = df_water.set_axis(range(df_water.shape[1]), axis=1)
df_fire_reset = df_fire.set_axis(range(df_fire.shape[1]), axis=1)
display(df_water_reset)
display(df_fire_reset)
print('\n**************************************************\n')

# Concatenate without automatic column alignment
result = pd.concat([df_water_reset, df_fire_reset], ignore_index=True, axis=0)
display(result.info())
display(result)
```

# 3. Pandas: Reshape the data

Reshape in NumPy vs. Pandas:
- In NumPy (`np.reshape()`):
  - `np.reshape()` is a low-level operation that changes the shape of an array without changing its data. It simply reinterprets the same block of memory with a new set of dimensions (= creates a new view).
  - The most common use cases involve preparing data for machine learning models or performing mathematical operations that require a specific matrix shape.
  - <u>It does not handle column names or row labels</u>.
- In Pandas:
  - Pandas' reshaping methods are high-level operations designed for data analysis, cleaning, and preparation.
  - They are about transforming the structure of data to make it more suitable for different types of analysis.
  - Pandas' reshaping methods build a new DataFrame by rearranging data based on its labels, often involving complex tasks like aggregation and handling of missing values.

In pandas, a tabular dataset can be written in two different formats:
1. **Wide format**
    - A wide format contains <u>values that **do not repeat** in a single column</u>.
    - For displaying data, we typically use a wide data format. Most datasets that we encounter in the real world will also be recorded in a wide format because it’s easier for our brains to interpret.
2. **Long format**
    - A long format contains <u>values that **do repeat** in a single column</u>.
    - During data processing and analyzing, the data should be in a long format. Also many functions might specifically require long-format data (e.g., when we’re visualizing multiple variables in a plot using statistical software, we typically must convert our data to a long format in order for the software to create the plot).

## 4.3 Conditional styling

Under-the-hood, the Styler object uses Cascading Style Sheets (CSS) to customise various parameters influencing the display of the DataFrame.
- This is done by passing style functions into the Styler object using `.apply()` or `.map()`.
  - `.map()` : apply a function to the DataFrame <u>element-wise</u>
  - `.apply(...,axis)` : apply a function to the DataFrame <u>row-wise, column-wise, or both</u> (depending on the `axis` parameter)
- <u>The output of style functions should therefore be strings containing CSS attribute-value pair</u>, in the `'attr: value'` format. If nothing is to be applied, the functions should return an empty string `''` or `None`.

### 4.3.1 Style cells that meet a condition

```python
# Style any cell that meets a condition

# Define a style function that takes each cell as an argument
def style_negative_cells( val ):
  # For any cell with negative numbers, change the text to red
  color = 'red' if (val < 0) else ''
  return f'color:{color};'

# Format and style the dataframe
df.style.format( formatter='{:.2f}' ) \
        .map( style_negative_cells )
```

```python
# Generalize the style function with input argument(s)
# Chain many style functions
# Style the selected columns (subset)

# Define a generalized style function (with the 'attr' argument)
def style_negative_cells( val, attr ):
  return attr if (val < 0) else ''

# Format and style the dataframe
# 1. Format all columns (set floating-point precisions)
# 2. Style cells with negative values (change text color)
# 3. Style cells (in the selected columns) whose values in the range (change opacity)
s1 = df.style.format( formatter='{:.2f}' ) \
             .map( style_negative_cells, attr='color:red;background-color:black;' ) \
             .map( lambda v: 'opacity:30%;' if (-0.5 < v < 0.5) else None , subset=['A','B'] )
s1
```

### 4.3.2 Style rows/columns that meet a condition

```python
# Style all rows that contain no negative number

# A Style function
def style_allpositive_rows( row, attr ):
  # Choose the color for this one row
  attr = attr if (row >= 0).all() else ''

  # Assign the same color to all columns in this row
  return [ attr for col in row ]

# Apply the function in a row-wise manner
s2 = s1.apply(style_allpositive_rows, attr='background-color:orange;', axis=1)
s2
```

```python
# Highlight the maximum value(s) regarding each column

# A Style function
def style_max(val, attr=''):
  # If a value equals the maximum (ignore NA), set it to 'attr'. Otherwise, set it to ''.
  return np.where(val == np.nanmax(val.values), attr, '')

# Use blue cells to highlight the maximum of the column
df.style.format( formatter='{:.2f}' ) \
        .map( style_negative_cells, attr='color:red;' ) \
        .apply(style_max, attr='color:white;background-color:darkblue', axis=0)
```

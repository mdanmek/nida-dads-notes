## 01_data_understanding_and_preparation.ipynb

### Purpose

เตรียมข้อมูลจัดซื้อจัดจ้างปีงบประมาณ 2569
และสร้างชุดข้อมูลจ้างก่อสร้างสำหรับนำไปวิเคราะห์ต่อ

### Scope

- อ่านไฟล์ต้นทางขนาดใหญ่แบบ chunk
- สำรวจจำนวนรายการตามประเภทโครงการ
- เลือกกลุ่ม `จ้างก่อสร้าง`
- สกัดข้อมูลจ้างก่อสร้างทุกคอลัมน์
- ตรวจจำนวนแถว จำนวนคอลัมน์ และ grain เบื้องต้น
- บันทึก `construction_contracts_2569.csv`

### Output

- ข้อมูลจ้างก่อสร้าง 180,079 แถว
- `construction_contracts_2569.csv`
- ข้อสรุปว่าควรวิเคราะห์ข้อมูลที่ระดับโครงการเป็นหลัก
  และใช้ข้อมูลระดับแถวเมื่อตรวจสอบสัญญาหรือผู้รับจ้าง

### Out of Scope

- การวิเคราะห์ distribution อย่างละเอียด
- การสร้าง Red Flag
- การจัดอันดับผู้รับจ้างหรือหน่วยงาน
- การสรุป Insight ของโครงการ

## 02_construction_eda.ipynb

### Purpose

ทำความเข้าใจลักษณะของโครงการจ้างก่อสร้าง
และค้นหารูปแบบสำคัญจากข้อมูลก่อนกำหนด Red Flag

### Scope

1. โหลดข้อมูลจ้างก่อสร้างที่เตรียมไว้
2. เลือกตัวแปรที่จะใช้ในการวิเคราะห์
3. ตรวจ missing values และชนิดข้อมูลเฉพาะคอลัมน์ที่ใช้
4. สร้างข้อมูลระดับโครงการโดยใช้ `รหัสโครงการ`
5. วิเคราะห์ภาพรวม ได้แก่
   - จำนวนและมูลค่าโครงการ
   - Distribution ของวงเงินงบประมาณ
   - ราคากลางและราคาที่ตกลง
   - วิธีการจัดซื้อจัดจ้าง
   - หน่วยงาน
   - จังหวัดหรือพื้นที่
6. สร้างตัวแปรพื้นฐาน เช่น
   - ส่วนต่างระหว่างงบประมาณกับราคาตกลง
   - ส่วนต่างระหว่างราคากลางกับราคาตกลง
   - ร้อยละของส่วนต่างราคา
7. วิเคราะห์ผู้ชนะการเสนอราคาและการกระจุกตัวเบื้องต้น
8. สรุป Findings และคำถามที่ควรเจาะลึกต่อ

### Main Questions

- โครงการจ้างก่อสร้างมีมูลค่ากระจายตัวอย่างไร
- ราคาตกลงแตกต่างจากงบประมาณและราคากลางเพียงใด
- รูปแบบราคาแตกต่างกันตามวิธีจัดซื้อหรือไม่
- หน่วยงานหรือพื้นที่ใดมีจำนวนและมูลค่าโครงการสูง
- ผู้รับจ้างมีการกระจุกตัวมากน้อยเพียงใด
- มีรูปแบบใดที่ควรนำไปตรวจสอบเพิ่มเติม

### Output

- ชุดข้อมูลระดับโครงการที่พร้อมวิเคราะห์
- ตารางและกราฟ EDA
- Findings ที่สนับสนุนด้วยตัวเลข
- Candidate Red Flags สำหรับ Notebook 03

### Out of Scope

- การกล่าวหาหรือยืนยันการทุจริต
- การกำหนด Red Flag Score ล่วงหน้า
- Machine Learning
- Predictive Model
- Dashboard
- Interactive Visualization
- การใช้เกณฑ์ anomaly ที่ยังไม่มีเหตุผลจาก EDA

## 03_red_flag_analysis.ipynb

### Purpose

นำ Findings จาก EDA มาสร้างเกณฑ์คัดกรองรายการ
ที่ควรได้รับการตรวจสอบเพิ่มเติม

### Scope

1. ทบทวน Candidate Red Flags จาก Notebook 02
2. นิยามและอธิบายเหตุผลของแต่ละ Red Flag
3. ตรวจจำนวนรายการที่เข้าแต่ละเงื่อนไข
4. วิเคราะห์ Red Flag แยกตาม:
   - วิธีจัดซื้อ
   - หน่วยงาน
   - ผู้รับจ้าง
   - ขนาดของโครงการ
5. ตรวจโครงการที่เข้าเงื่อนไขหลายข้อ
6. เลือกกรณีศึกษาที่น่าสนใจ
7. สรุปข้อจำกัดและเสนอ Action Plan

### Possible Red Flags

รายการต่อไปนี้เป็นเพียงแนวคิดเบื้องต้น
ต้องได้รับการสนับสนุนจากผล EDA ก่อนนำมาใช้จริง:

- ราคาตกลงใกล้ราคากลางมากผิดสังเกต
- ราคาตกลงสูงกว่างบประมาณหรือราคากลาง
- มูลค่าโครงการแตกต่างจากโครงการในกลุ่มเดียวกันมาก
- ผู้รับจ้างได้รับหลายโครงการหรือมีมูลค่ารวมสูง
- โครงการเดียวมีหลายสัญญาหรือหลายผู้ชนะ
- ข้อมูลราคา วันที่ หรือสัญญาไม่สอดคล้องกัน

### Output

- คำจำกัดความของ Red Flag แต่ละประเภท
- ตารางรายการที่ควรตรวจสอบเพิ่มเติม
- กรณีศึกษาที่ตรวจสอบย้อนกลับได้
- ข้อเสนอแนะและ Action Plan
- ข้อจำกัดของการวิเคราะห์

### Important Limitation

Red Flag หมายถึงสัญญาณที่ควรตรวจสอบเพิ่มเติม
ไม่ใช่หลักฐานยืนยันว่ามีการทุจริต
